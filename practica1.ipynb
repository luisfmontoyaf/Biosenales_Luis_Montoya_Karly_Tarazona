**REPRESENTACIÓN DE SEÑALES DISCRETAS**



```python
#- Impulso unitario:
def impseq(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n-n0) == 0
    return [x,n]
```

p) Implemente una función en Python que permita generar una función escalón unitario, definida en un intervalo n_1≤n_0≤n_2



```python
def escalon(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n-n0) == 0
    return [x,n]
```

```python
import numpy as np

def rampseq(n0, n1, n2, m):

    n = np.arange(n1, n2 + 1)
    x = m * (n - n0) * ((n - n0) >= 0)

    return x, n
```
