**REPRESENTACIÓN DE SEÑALES DISCRETAS**



```python
#- Impulso unitario:
def impseq(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n-n0) == 0
    return [x,n]
```

p) Implemente una función en Python que permita generar una función escalón unitario, definida en un intervalo n_1≤n_0≤n_2



```python
def escalon(n0, n1, n2):
#   Genera x(n) = delta(n-n0); n1 <= n <= n2
#	 ----------------------------------------------
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n-n0) >= 0
    return [x,n]
```

```python
import numpy as np

def rampseq(n0, n1, n2, m):

    n = np.arange(n1, n2 + 1)
    x = m * (n - n0) * ((n - n0) >= 0)

    return x, n
```

Con la función creada se puede sumar una función desplazada con otra sin desplazar.

```python

def sigadd(x1, n1, x2, n2):
    # 1. Definir el eje de tiempo global que cubra ambas señales
    n = np.arange(min(min(n1), min(n2)), max(max(n1), max(n2)) + 1)

    # 2. Crear vectores de ceros del tamaño total
    y1 = np.zeros(len(n))
    y2 = np.zeros(len(n))

    # 3. Insertar las amplitudes en sus posiciones temporales correctas
    y1[np.isin(n, n1)] = x1
    y2[np.isin(n, n2)] = x2

    # 4. Sumar las señales perfectamente alineadas
    y = y1 + y2
    return y, n

y,n_t = sigadd(x,n,x_desplazado,n5)

fig = plt.figure(figsize=(10, 3)) 
plt.plot(n_t, y)
plt.title("Señal según su salida")
plt.xlabel("N") 
plt.ylabel("Amplitud")
plt.legend()
plt.grid()
```

1)	Escribir una función que reciba un diccionario con las notas de los alumnos de un curso y devuelva una serie con la nota mínima, la máxima, media y la desviación típica. 

```python
import pandas as pd

def estadisticas_notas(notas):
    serie_notas = pd.Series(notas)

    resultado = pd.Series({
        "Nota mínima": serie_notas.min(),
        "Nota máxima": serie_notas.max(),
        "Media": serie_notas.mean(),
        "Desviación típica": serie_notas.std()
    })

    return resultado


# Diccionario con las notas
notas = {
    "Ana": 4.5,
    "Juan": 3.8,
    "Carlos": 4.2,
    "María": 3.5,
    "Pedro": 4.0
}

print(estadisticas_notas(notas))
```

a)	 Cargar los datos desde el archivo "datos.csv".

d)	 Calcular el IMC con la fórmula:
IMC=  (Peso (kg))/〖(Altura (m))〗^2 

e)	 Agregar el IMC como una nueva columna llamada "BMI" en el DataFrame.

f)	 (Opcional): Categorizar a los participantes según su IMC 


```python
import pandas as pd

# a) Cargar los datos desde el archivo datos.csv
df = pd.read_csv("datos.csv", sep=";")

print("Columnas del archivo:")
print(df.columns)

# b) Mostrar las primeras y últimas filas
print("Primeras filas:")
print(df.head())

print("\nÚltimas filas:")
print(df.tail())

# c) Borrar la columna "Unnamed: 0"
df_act= df.drop("Unnamed: 0", axis=1) 
print("Columnas del archivo:")
print(df_act.columns)


# d) Calcular el IMC
# Altura está en centímetros, por eso se divide entre 100
df_act["BMI"] = (df["Weight"]) / (((df["Height"]) / 100) ** 2)


#e) Mostrar el DataFrame con la nueva columna BMI
print("\nDataFrame con IMC:")
print(df_act)
```

f)	 (Opcional): Categorizar a los participantes según su IMC en:

```python
# f) Categorizar según el IMC

def categoria_imc(bmi):
    if bmi < 18.5:
        return "Bajo peso"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Sobrepeso"
    else:
        return "Obesidad"

df_act["Categoria"] = df_act["BMI"].apply(categoria_imc)

print("\nDataFrame con categoría de IMC:")
print(df_act)
```
