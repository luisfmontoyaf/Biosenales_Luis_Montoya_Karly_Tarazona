1. Realice un esquema donde se describan y expliquen las diferentes etapas del registro de biopotenciales, desde la generación de los potenciales de acción hasta su adquisición por el equipo. Adicionalmente, describa a grandes rasgos los componentes de un equipo de adquisición.

Células excitables → Potencial de acción → Sumación espacio-temporal 
→ Propagación en el conductor volumétrico (tejidos) → Interfaz piel-electrodo (transducción iónica→electrónica)
→ Preamplificador de instrumentación → Filtrado / amplificación → ADC → Procesamiento y visualización.

1. Generación: Las células excitables (cardiomiocitos, fibras musculares, neuronas) mantienen un potencial de reposo de entre −70 y −90 mV gracias a la bomba Na⁺/K⁺ y a la permeabilidad selectiva de la membrana. Un estímulo supraumbral abre canales de Na⁺ (despolarización) y luego de K⁺ (repolarización).
2. Sumación y propagación: En el corazón, el PA (potencial de accion) nace en el nodo sinoauricular y se propaga por aurículas,luego va al nodo AV, haz de His y fibras de Purkinje hasta los ventrículos. En el músculo, un PA de la motoneurona activa todas las fibras de su unidad motora (potencial de acción de la unidad motora, MUAP). La señal medida es la suma de muchos PA asincrónicos (EMG) o sincrónicos (ECG).
3. Conducción en el volumen conductor: Las corrientes iónicas se distribuyen por los tejidos (grasa, músculo, piel), que actúan como filtro pasa-bajos espacial lo que significa que el filtrado depende de la distancia y no del tiempo, por eso la señal en superficie es de microvoltios a pocos milivoltios, frente a ~100 mV dentro de la célula.
4. Transducción en el electrodos: El electrodo de Ag/AgCl con gel electrolítico convierte la corriente iónica del cuerpo en corriente electrónica (reacción Ag + Cl⁻ ⇌ AgCl + e⁻). Aquí aparecen el potencial de media celda y la impedancia piel-electrodo.
5. Acondicionamiento: Un amplificador de instrumentación diferencial rechaza el ruido de modo común (CMRR alto) y amplifica la diferencia entre electrodos; filtros pasa-altos eliminan el offset DC y la deriva, y filtros pasa-bajos limitan el ancho de banda y evitan aliasing. 
6. Digitalización: El ADC no ve la señal de forma continua, sino que toma capturas cada cierto tiempo. La frecuencia de muestreo (fs). Para no perder información, fs debe ser al menos el doble de la frecuencia más alta de la señal. Si no se cumple, aparece el aliasing: frecuencias altas se pasan por frecuencias bajas y distorsionan el registro. 
7. Procesamiento y presentación: Filtrado digital, detección de eventos (picos R, contracciones), extracción de características, almacenamiento y visualización.

3. Describa ¿Cuál es la problemática asociada al acoplamiento piel-electrodo? ¿Cuáles estrategias se utilizan para mejorarla? Además, responda ¿Por qué el acoplamiento piel-electrodo constituye la etapa más crítica en el registro de señales de superficie (sEMG, EEG, ECG)?

La capa más externa de la piel (estrato córneo) es mala conductora, la impedancia piel-electrodo puede ir de decenas de kΩ a más de 1 MΩ a baja frecuencia, y se modela como un circuito RC en paralelo en serie con la resistencia del electrolito; esto puede causar:

•	Pérdida de señal por divisor de tensión entre la impedancia del electrodo y la impedancia de entrada del amplificador.

•	Desbalance de impedancias entre electrodos, que convierte el ruido de modo común (60 Hz) en ruido diferencial y empeora el CMRR efectivo.

•	Potenciales de offset y deriva por la interfaz electrodo-electrolito y por la sudoración.

•	Artefactos de movimiento: al estirar la piel varía el potencial de la capa córnea, generando variaciones de hasta varios mV, comparables o mayores que la señal de interés.

•	Variabilidad en el tiempo, porque el gel se seca y la piel se hidrata o suda.

**Estrategias de mejora**

•	Limpiar con alcohol, rasurar y, si es necesario, abrasionar levemente la piel para reducir el estrato córneo.

•	Usar electrodos Ag/AgCl con gel electrolítico, de buena calidad y del mismo tipo/lote.

•	Fijarlos bien y evitar tensión del cable; esperar unos minutos a que se estabilice la interfaz.

•	Amplificadores con alta impedancia de entrada, electrodos activos (preamplificador junto al electrodo), circuito DRL y cables apantallados.

•	Verificar la impedancia de contacto antes del registro y mantenerla baja y balanceada.

**¿Por qué es la etapa más crítica?**

 Porque es el primer eslabón de la cadena; cualquier ruido, offset o pérdida de señal que se introduzca allí se amplifica junto con la señal y no puede separarse después. El ruido propio de los electrodos limita la resolución de los registros de biopotenciales (Godin et al., 1991; Huigen et al., 2002). Además, las señales de superficie son muy pequeñas (µV en EEG, 0,05–5 mV en sEMG y ECG) , mientras que los artefactos del contacto, como el offset del electrodo o el movimiento, son del mismo orden de magnitud o mayores.(Webster, 1984; Webster, 2009).Por último, cuando la impedancia piel-electrodo es alta o desigual entre electrodos, la interferencia de red de modo común se convierte en ruido diferencial, y el rechazo de modo común efectivo del sistema es menor que el del amplificador (Huhta y Webster, 1973; Pallás-Areny, 1988; Wood et al., 1995), lo que empeora la relación señal-ruido.


4. Análisis de la Variabilidad de la frecuencia cardíaca (HRV): 

- Detectar los picos R utilizando la función de la librería NeuroKit y de Detectors. ¿Cuál de los algoritmos detecta mejor los picos R?

- Calcular la variabilidad de la frecuencia cardíaca (HRV) a partir de los picos R detectados.

- Investigar e implementar las siguiente funcion de la librería hrvanalysis para extraer características de HRV en el dominio del tiempo: get_time_domain_features

```python
# -*- coding: utf-8 -*-
"""
Análisis de ECG: detección de picos R, comparación de detectores y HRV.

"""
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import neurokit2 as nk
from scipy.signal import butter, filtfilt
from ecgdetectors import Detectors

warnings.filterwarnings('ignore')

# ----------------------------------------------------------------------
# 1) PARÁMETROS
# ----------------------------------------------------------------------
ARCHIVO = 'EMG.csv'          # ruta de su archivo (columna con encabezado 'EMG')
COLUMNA = 'EMG'
FS = 100                     # Hz: delay(10) en Arduino -> ~100 Hz
TOL_MS = 50                  # tolerancia para comparar picos (ms)

# Picos R de referencia (muestras), verificados en la gráfica.
REF = np.array([100, 178, 266, 365, 457, 547, 645, 759, 862, 960])

# ----------------------------------------------------------------------
# 2) LECTURA Y LIMPIEZA
# ----------------------------------------------------------------------
x = pd.read_csv(ARCHIVO)[COLUMNA].values.astype(float)

# Valores erróneos del arranque.
MALAS = [0, 13]
x[MALAS] = np.nan
ecg = pd.Series(x).interpolate(limit_direction='both').values
ecg = ecg - np.mean(ecg)                    # quitar offset DC
t = np.arange(len(ecg)) / FS                # tiempo en segundos

# Filtro pasa-banda 0,5–40 Hz 
b, a = butter(3, [0.5, 40], btype='band', fs=FS)
clean = filtfilt(b, a, ecg)

# ----------------------------------------------------------------------
# 3) GRÁFICAS PARA VERIFICAR LOS PICOS 
# ----------------------------------------------------------------------
plt.figure(figsize=(20, 5))
plt.plot(ecg)
plt.xticks(range(0, 1001, 25)); plt.grid(alpha=0.3)
plt.show()

# ----------------------------------------------------------------------
# 4) DETECCIÓN DE PICOS R CON VARIOS ALGORITMOS
# ----------------------------------------------------------------------
picos = {}

# Métodos de NeuroKit2 (sobre la señal filtrada)
for m in ['neurokit', 'pantompkins1985', 'hamilton2002',
          'christov2004', 'elgendi2010']:
    try:
        _, info = nk.ecg_peaks(clean, sampling_rate=FS, method=m)
        picos['NK-' + m] = np.array(info['ECG_R_Peaks'])
    except Exception as e:
        print(f'NK-{m} no se pudo usar con FS = {FS} Hz:', e)

# Detectores de py-ecg-detectors (sobre la señal sin filtrar)
det = Detectors(FS)
detectores = {
    'DET-pan_tompkins': det.pan_tompkins_detector,
    'DET-hamilton':     det.hamilton_detector,
    'DET-christov':     det.christov_detector,
    'DET-engzee':       det.engzee_detector,       # falla con FS = 100 Hz
    'DET-two_average':  det.two_average_detector,
}
for nombre, funcion in detectores.items():
    try:
        picos[nombre] = np.array(funcion(ecg))
    except ValueError as e:
        print(f'{nombre} no se puede usar con FS = {FS} Hz:', e)

# ----------------------------------------------------------------------
# 5) COMPARACIÓN CONTRA LA REFERENCIA
# ----------------------------------------------------------------------
tol = int(TOL_MS / 1000 * FS)               # 5 muestras con FS = 100


def evaluar(det_idx, ref_idx, tol):
    det_idx = np.array(det_idx)
    tp = sum(np.any(np.abs(det_idx - r) <= tol) for r in ref_idx)
    fn = len(ref_idx) - tp
    fp = sum(not np.any(np.abs(ref_idx - d) <= tol) for d in det_idx)
    sens = tp / (tp + fn) if (tp + fn) else 0
    ppv = tp / (tp + fp) if (tp + fp) else 0
    return tp, fp, fn, sens, ppv


filas = []
for nombre, p in picos.items():
    tp, fp, fn, sens, ppv = evaluar(p, REF, tol)
    filas.append([nombre, len(p), tp, fp, fn, round(sens, 2), round(ppv, 2)])

tabla = pd.DataFrame(filas, columns=['Detector', 'n_picos', 'TP', 'FP', 'FN',
                                     'Sensibilidad', 'PPV'])
tabla = tabla.sort_values(['Sensibilidad', 'PPV'], ascending=False)
print('\n=== Comparación de detectores (tolerancia ±%d ms) ===' % TOL_MS)
print(tabla.to_string(index=False))

# ----------------------------------------------------------------------
# 6) HRV EN EL DOMINIO DEL TIEMPO (con los picos de referencia)
# ----------------------------------------------------------------------

import astropy.stats
from astropy.timeseries import LombScargle
astropy.stats.LombScargle = LombScargle
from hrvanalysis import (remove_outliers, interpolate_nan_values,
                         remove_ectopic_beats, get_time_domain_features)
 

picos_hrv = REF
 
rr_ms = list(np.diff(picos_hrv) / FS * 1000)          # intervalos RR en ms
 
# Limpieza de la serie RR
rr_ms = remove_outliers(rr_intervals=rr_ms, low_rri=300, high_rri=2000)
rr_ms = interpolate_nan_values(rr_intervals=rr_ms, interpolation_method='linear')
nn = remove_ectopic_beats(rr_intervals=rr_ms, method='malik')   # latidos ectópicos
nn = interpolate_nan_values(rr_intervals=nn)                    # serie NN final
 
# Características de HRV en el dominio del tiempo
features = get_time_domain_features(nn)
print('\n=== HRV en el dominio del tiempo (hrvanalysis) ===')
for k, v in features.items():
    print(f'{k:<12}: {v:.2f}')
 
rr = np.array(nn)        # se usa en el tacograma
 
# ----------------------------------------------------------------------
# 7) FIGURAS FINALES
# ----------------------------------------------------------------------
plt.figure(figsize=(18, 5))
plt.plot(t, ecg, lw=1, label='ECG')
plt.plot(t[picos_hrv], ecg[picos_hrv], 'ro', label='Picos R')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud (cuentas ADC)')
plt.xticks(np.arange(0, t[-1] + 0.5, 0.5))
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig('ecg_picos.png', dpi=200)
plt.show()
 
plt.figure(figsize=(8, 4))
plt.plot(np.arange(1, len(rr) + 1), rr, 'o-')
plt.xlabel('Latido')
plt.ylabel('Intervalo RR (ms)')
plt.title('Tacograma')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('tacograma.png', dpi=200)
plt.show()

```

# Aplicaciones clínicas del ECG y el EMG

## A. Aplicaciones del electrocardiograma (ECG)

### 1. Detección de arritmias cardíacas

Permite identificar alteraciones del ritmo, como fibrilación auricular y taquicardia. Se utiliza un ECG de 12 derivaciones o un monitor Holter para registros prolongados. El procesamiento incluye filtrado digital, detección de complejos QRS y cálculo de intervalos RR para determinar la frecuencia y regularidad cardíaca (American Heart Association, s. f.; Xie et al., 2020).

### 2. Detección de isquemia e infarto de miocardio

Permite identificar cambios eléctricos asociados con lesiones cardíacas. Se utiliza un ECG de 12 derivaciones en reposo. La señal se acondiciona para reducir el ruido y se analizan las ondas, el segmento ST y sus variaciones para identificar alteraciones compatibles con isquemia o infarto (Byrne et al., 2023).

### 3. Pruebas de esfuerzo cardíaco

Evalúan la respuesta del corazón durante el ejercicio. Se registra continuamente el ECG mientras el paciente camina en una banda sin fin o pedalea en una bicicleta estática. Se analizan la frecuencia cardíaca, el ritmo y los cambios del segmento ST durante el esfuerzo y la recuperación (American Heart Association, s. f.).

## B. Aplicaciones de la electromiografía (EMG)

### 4. Diagnóstico de enfermedades neuromusculares

Permite estudiar lesiones nerviosas y enfermedades musculares. Se utiliza EMG de aguja, registrando la actividad muscular en reposo y durante la contracción. Se evalúan la amplitud, duración y morfología de los potenciales de las unidades motoras, así como la actividad espontánea anormal (Cleveland Clinic, s. f.).

### 5. Evaluación de la marcha y rehabilitación

Permite analizar la activación muscular durante el movimiento, por ejemplo, en pacientes con hemiparesia después de un accidente cerebrovascular. Se emplea EMG de superficie, con electrodos ubicados sobre la piel. El procesamiento incluye filtrado, rectificación, cálculo de la envolvente y análisis de amplitud y duración de la activación muscular (Chowdhury et al., 2013; Pradon et al., 2024).

### 6. Control de prótesis mioeléctricas

Permite controlar movimientos protésicos mediante las contracciones musculares de personas con amputaciones. Se utilizan varios canales de EMG de superficie. La señal se filtra, se divide en ventanas y se extraen características como el valor absoluto medio (MAV) y el valor cuadrático medio (RMS). Después, un algoritmo clasifica los patrones musculares y los convierte en órdenes para controlar la prótesis (Chowdhury et al., 2013; Geethanjali, 2016).


## Conclusiones

1. En la adquisición de potenciales biológicos se tiene una cadena de transducción donde los fenómenos iónicos celulares se transforman de manera física y electrónica para llegar a ser información que podamos usar en el ámbito clínico. Los tejidos biológicos son barreras que actúan como filtros, ya que no permiten el paso completo de la señal. Adicionalmente, la interfaz piel-electrodo incluye en el sistema variables electroquímicas. Para obtener un diagnóstico correcto, se busca evitar errores en cada eslabón del proceso. Para esto, se necesita un hardware que permita una gran resolución en la medición de potenciales muy pequeños y también una correcta digitalización que evite distorsiones que no puedan revertirse.

2. El ECG y el EMG son herramientas fundamentales para el diagnóstico y seguimiento de diferentes enfermedades. Permiten analizar la actividad eléctrica del corazón y de los músculos, respectivamente. La utilidad clínica depende de varios aspectos, como el tipo de registro que se utiliza y el procesamiento que se le aplica a la señal. En el ECG, técnicas como el filtrado y la detección de complejos QRS permiten identificar alteraciones del ritmo y cambios compatibles con lesiones cardíacas. En el EMG, el procesamiento permite evaluar aspectos de las enfermedades neuromusculares, el movimiento de la persona e incluso el control de una prótesis mioeléctrica.

3. El límite en la calidad de la adquisición de una señal biológica, como el EMG o el ECG, está determinado en gran medida por la interfaz piel-electrodo. En este punto de contacto entre el electrodo y la piel, cualquier ruido introducido puede amplificarse junto con los pequeños potenciales de las señales capturadas, afectando la calidad de la medición. Para superar la alta impedancia del estrato córneo, se necesita una preparación física del paciente (limpieza y abrasión de la piel, así como el uso de electrodos Ag/AgCl de calidad) y hardware adecuado. Ambas medidas ayudan a evitar el ruido que deteriora la señal desde la adquisición y que puede persistir incluso después del procesamiento digital.
